```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
def image_show(title, img):
    plt.rcParams['figure.figsize'] = [10, 5]
    plt.title(title)
    plt.imshow(img)
    plt.show()

img=cv2.imread('imagen1.jpg')
imgRGB=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
imgGray=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imwrite("imagenBGR.jpg", img) #Guardados
cv2.imwrite("imageRGB.jpg", imgRGB)
cv2.imwrite("imageGRAY.jpg", imgGray)
image_show("imagen en escala de grises", imgGray)

plt.subplot(131)
plt.title("Imagen BGR")
plt.imshow(img)

plt.subplot(132)
plt.title("Imagen RGB")
plt.imshow(imgRGB)

plt.subplot(133)
plt.title("Imagen en escala de grises")
plt.imshow(imgGray, cmap='gray')
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(img)
axs[0].set_title("imagen BGR")
axs[1].imshow(imgRGB)
axs[1].set_title("imagen RGB")
axs[2].imshow(imgGray, cmap='gray')
axs[2].set_title("imagen en escala de grises")

```
