```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['figure.figsize'] = [10, 5]
size_imagen=100
imagen = np.ones((size_imagen, size_imagen, 3), dtype=np.uint8) * 255

mitad = size_imagen // 2

for i in range(size_imagen):
    for j in range(size_imagen):
        if i < mitad and j < mitad:
            imagen[i, j] = [255, 0, 0]
        elif i < mitad and j >= mitad:
            imagen[i, j] = [0, 255, 0]
        elif i >= mitad and j < mitad:
            imagen[i, j] = [0, 0, 255]
        else:
            imagen[i, j] = [255, 255, 0]
plt.figure()
plt.imshow(imagen)
plt.title("Imagen con cuatro cuadrantes de color distinto")
plt.show()
```
