Roberts mejorado

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def filtro_roberts_mejorado(img):
    img_gris=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    suavizar=cv2.GaussianBlur(img_gris,(3,3),0)
    robertx=np.array([[-2,0], [0,2]],dtype=np.float32)
    roberty=np.array([[0,-2], [2,0]],dtype=np.float32)
    r_x=cv2.filter2D(suavizar,-1,robertx)
    r_y=cv2.filter2D(suavizar,-1,roberty)
    roberts=cv2.addWeighted(np.abs(r_x),0.7,np.abs(r_y),0.7,0)
    _,bordes=cv2.threshold(roberts,20,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)
    return bordes

img=cv2.imread('imagen2.jpg')
bordes=filtro_roberts_mejorado(img)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')
plt.subplot(1,2,2)
plt.imshow(bordes, cmap='gray')
plt.title('Bordes con Filtro de Roberts Mejorado')
```

Prewitt Mejorado

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def filtro_prewitt_mejorado(img):
    img_gris=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    suavizar=cv2.GaussianBlur(img_gris,(3,3),0)
    kernel_x=np.array([[-3,0,3], [-3,0,3], [-3,0,3]],dtype=np.float32)
    kernel_y=np.array([[3,3,3], [0,0,0],[-3,-3,-3]],dtype=np.float32)
    pre_x=cv2.filter2D(suavizar,cv2.CV_64F,kernel_x)
    pre_y=cv2.filter2D(suavizar,cv2.CV_64F,kernel_y)
    magnitud=cv2.sqrt(pre_x**2 + pre_y**2)
    _,bordes=cv2.threshold(magnitud,255,255,cv2.THRESH_BINARY)
    return bordes

img=cv2.imread('imagen2.jpg')
bordes=filtro_prewitt_mejorado(img)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')
plt.subplot(1,2,2)
plt.imshow(bordes, cmap='gray')
plt.title('Bordes con Filtro de Prewitt Mejorado')
```

Filtro canny

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def filtro_canny(img,umbral_bajo=100,umbral_alto=200):
    img_gris=cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    suavizar=cv2.GaussianBlur(img_gris,(3,3),0)
    bordes=cv2.Canny(suavizar,umbral_bajo,umbral_alto)
    return bordes

img=cv2.imread('imagen2.jpg')
bordes=filtro_canny(img)
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')
plt.subplot(1,2,2)
plt.imshow(bordes, cmap='gray')
plt.title('Bordes con Filtro de Canny')
```
