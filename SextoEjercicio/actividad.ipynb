```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convulacion_media_h(imagen,tkernel):
    kernel = np.ones((1,tkernel),np.float32)/tkernel
    imagen_c=cv2.filter2D(imagen,-1,kernel)
    return imagen_c

img=cv2.imread('descarga.jpg')
img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
img_horizontal=convulacion_media_h(img_rgb,20)
factor = 10
img_aliasing = img_horizontal[::factor, ::factor]
img_aliasing_up = cv2.resize(img_aliasing, (img.shape[1], img.shape[0]), interpolation=cv2.INTER_NEAREST)


plt.figure(figsize=(10,10))
plt.subplot(321)
plt.imshow(img_rgb)
plt.title('Imagen Original')
plt.subplot(322)
plt.imshow(img_horizontal)
plt.title('Imagen con filtro de media horizontal')
plt.subplot(323)
plt.imshow(img_aliasing)
plt.title('Imagen con aliasing')
ct_x = np.linspace(0.0, 1.0, 1024)
ct_y = np.sin(ct_x * 4 * np.pi)
ax = plt.subplot(3, 2, 4)
ax.set_xlim(0, 1)
ax.set_yticks(np.linspace(-2 ,2, 5))
ax.set_ylim(-2, 2)
ax.grid(color='k', linestyle='--')
ax.set_title("Señal normal")
ax.plot(ct_x, ct_y, 'k')
plt.show()
```
