```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img=cv2.imread('imagen2.jpg',0 )
kernel=np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])#mascara de convulacion
img_fin=cv2.filter2D(img,-1,kernel)
plt.figure(figsize=(10,5))
plt.subplot(121)
plt.imshow(img,cmap='gray')
plt.subplot(122)
plt.imshow(img_fin,cmap='gray')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img=cv2.imread('imagen2.jpg')
img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
kernel=np.ones((5,5),np.float32)/25
suavizado=cv2.filter2D(img_rgb,-1,kernel)
kernel2=np.ones((9,9),np.float32)/81
suavizado2=cv2.filter2D(img_rgb,-1,kernel2)
plt.figure(figsize=(15,5))
plt.subplot(131)
plt.imshow(img_rgb)
plt.subplot(132)
plt.imshow(suavizado)
plt.subplot(133)
plt.imshow(suavizado2)
plt.show()
```
