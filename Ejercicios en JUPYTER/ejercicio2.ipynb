```python
s = input('Introduce un número entero: ')
n = int(s)
if primo(n):
    print(f'{n} es primo')
else:
    print(f'{n} no es primo')
def primo(n):
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    import math
    r = int(math.isqrt(n))
    for i in range(3, r + 1, 2):
        if n % i == 0:
            return False
    return True

```
