```python
numeros = []
for i in range(7):
    num = int(input(f'Introduce el número {i+1}: '))
    numeros.append(num)
repeticiones = {}
for num in numeros:
    if num in repeticiones:
        repeticiones[num] += 1
    else:
        repeticiones[num] = 1
numero_mas_repetido = max(repeticiones, key=repeticiones.get)
veces_repetido = repeticiones[numero_mas_repetido]
    
print(f'\nNúmeros ingresados: {numeros}')
print(f'\nEl número que más se repite es: {numero_mas_repetido}')
print(f'Se repite {veces_repetido} veces')
```
